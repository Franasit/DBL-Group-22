# Flood Dynamics in South Sudan: End-to-End Master Pipeline
This notebook executes the complete data science pipeline from **raw datasets** (satellite flood masks, lake levels, river discharge, gridded population, cattle rasters, cropland masks, and health facility coordinates) to **exploratory data analysis (EDA)**, **spatial risk mapping**, **Random Forest machine learning**, and **16-day future flood risk forecasting**.

### Pipeline Stages
1. **Upstream Hydrological Analysis**: Long-term water levels (Lakes Victoria, Kyoga, Albert) and Nile river discharge trends.
2. **Spatial Data Cleaning & Multi-Source Alignment (ADM3 Payam Level)**: Extraction of zonal statistics from rasters and spatial joins with health facilities and flood masks.
3. **Exposure Normalization & Composite Scoring**: Min-max normalization and combined exposure indexes.
4. **Two-Factor Exposure Analysis (EDA)**: Factor 1 (Human/Pastoral) vs. Factor 2 (Agronomic/Healthcare Infrastructure).
5. **Spatial Risk Mapping & Infrastructure Validation**: Geographic exposure mapping of health infrastructure across South Sudan.
6. **Predictive Modeling with Random Forest**: 5-fold Stratified GridSearchCV for flood risk classification.
7. **Future Flood Risk Forecasting**: Integrating the trained Random Forest model with live Open-Meteo 16-day precipitation forecasts.
